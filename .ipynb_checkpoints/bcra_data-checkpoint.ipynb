```python
import pandas as pd
import requests
import matplotlib.pyplot as plt
!pip install jupyterthemes
```

```python
def traer_serie_bcra():
    url = 'https://api.bcra.gob.ar/estadisticas/v3.0/monetarias/1?desde=2024-01-01&hasta=2025-04-25&limit=1000'
    r = requests.get(url, verify=False)

    if r.status_code == 200:
        datos = r.json()
        
        resultados = datos['results']

        fechas = []
        valores = []

        for item in datos['results']:
            fechas.append(item['fecha'])
            valores.append(item['valor'])

        fechas = pd.to_datetime(fechas)

        plt.figure(figsize=(15,4))
        plt.plot(fechas, valores, marker='o', color='g', linestyle='-', markersize='4')

        plt.title('Reservas Netas del BCRA', fontsize=15)
        plt.xlabel('Periodo', fontsize=12)
        plt.ylabel('Valor en millones de pesos', fontsize=12)
        
        plt.xticks(rotation=45)
        plt.tight_layout()
        
        plt.show()
        return datos
    else:
        print(f"Error al obtener los datos: {r.status_code}")
        return None
    


traer_serie_bcra()
```
